# **Context**
I decided to do my AI-B project on 'MS Gen 13', which is a handwritten 19th century travel journal held at the University of Glasgow's Archives and Special Collections (ASC). Andrew MacGeorge, a Glaswegian author and historian, wrote this journal during a continental tour he took between late 1867 and early 1868. The journal itself comes to just under 20,000 words and is a detailed account of Victorian era "life on the road", documenting hotels to railways, food to weather, and everything in between. A full transcription of the text was produced by Robert MacLean from the ASC, as well as all pages from the journal itself being digitised.

<br>

Due to the limitations of photographic technology at the time, capturing spontaneous snapshots of everyday life just wasn't feasible, yet many moments in this journal deserved to be preserved. This journal makes it clear that MacGeorge had a keen eye for observation, which is why, through his detailed descriptions and reflections throughout the journey, I've attempted to "bring them to life" so to speak. Within the journal are various sketches and even some photographs, however they're as you would expect for the 19th century: sepia, poor dynamic range, and in general, just not the best photographs to get a real impression of the locations where the photographs were captured. Photography during that period required stillness from people in the frame (due to long exposure), good lighting (hence the popularity of photography done in studios), and a camera that was already prepared to actually take the picture. For this reason many of the moments outlined in the journal just wouldn't be possible to capture on cameras of that time period, therefore it can be assumed his sketches throughout the journal were his way of capturing moments where a camera simply couldn't. With that said, not many of the sketches or photographs seem to capture the mundane. Not the mundane as in "boring", but mundane as in the things we now take for granted from having a camera constantly in our pockets, such as the snapshots of food before we eat, or the vistas we see on holiday. We think nothing of it now, but back then this all had to be planned.

<br>

This project stems from a simple observation which is that spontaneity in photography during the Victorian era more or less didn't exist. This came from looking at pictures of my own ancestors, clearly in photography studios, and thinking how fascinating it would have been to have seen them in a more "everyday life" type setting. Photography back then seemed more like a formal event than anything else, as a subject had to remain as still as possible, while sitting in front of a camera that took time to set up for each photograph, within the studio itself which had to have adequate enough lighting to actually capture the photograph... in other words, spontaneity and photography at the time were simply incompatible.

<br>

The Polaroid camera [as we now think of it](https://en.wikipedia.org/wiki/Polaroid_Land_Camera_1000), released just over a century after MacGeorge's journal, changed the photography landscape completely as now taking a photo could be instant, portable, and most importantly informal. Now you could capture moments as they happened without worrying if what you've just captured was actually "worth capturing", in other words, no more tedious preparation for both the photographer and subject. If MacGeorge were to have had a Polaroid camera with him throughout his journey, it's no question his journal would have been an even more fascinating documentation of travel throughout that period of history. Hence my topic for this project: *what would MacGeorge's journey look like if he had the ability to take pictures wherever and whenever he wished?*

# **Methodology**
I approached this project as follows:
1. Selected moments from the journal for the Polaroids.
2. Iterated on prompts to achieve the intended aesthetic.
3. Generated the Polaroids with ChatGPT image generation.
4. Wrote individual captions for each Polaroid in collaboration with ChatGPT.

### 1. Selecting 15 Polaroid Candidates
As this journal is approximately 20,000 words, I didn't have time to read through it all, therefore I decided to run the text transcription through ChatGPT and ask it to provide me with an extensive list of candidates for Polaroids which were interestingly mundane, in a way that they were everyday life moments that might make us reach for our phones in this day and age if we were to experience them. Eventually a list of 40 candidates were produced. I wanted to make sure these candidates covered the journey nicely in terms of geographic distribution, so from that original 40 I whittled them down to 15 which seemed "Polaroid Worthy".

### 2. Prompt Design and Iteration
As someone that's not used LLMs or image generation much, this was all a very interesting learning experience, as I quickly learned image generation is extremely finicky and hard to keep consistent.

<br>

**Iteration 1: Reference Template Image**
I originally planned on having ChatGPT generate me a "template" polaroid to work from, whereby the paper/card itself could be reused, and just the image on it could be changed for each of the 15 candidates. However, this ended up producing consistently poor quality results. ChatGPT would just apply the image to the template without much care or thought as to making it look like a real polaroid, instead of a quick Photoshop mockup. To add to this, it also had a habit of clipping/cropping the edges of the Polaroid itself, so the Polaroids weren't ever properly centred in the frame.

**Iteration 2: Overly Detailed Prompts**
My next approach was to generate images purely from text, thinking that if I were to just make a detailed outline of everything that needed to be captured right down to the specifics, I'd end up with my desired result. I included things such as specifying that the caption should be "written with a fountain pen, in 19th century cursive handwriting, making sure the cursive handwriting style was contemporary to the period". Perhaps I just needed more time to experiment, but from the attempts of using this granular approach to prompts, it took forever for the images to generate, if at all, as mostly ChatGPT just hung and didn't produce an output. So my takeaway from this was that overly detailed prompts were just as counterproductive as stripped back simplified prompts.

**Iteration 3: Template File System**
After taking a while to think about these issues, I finally landed on a method that produced consistency with no mistakes, whereby I thought modularly instead of just creating 1 gigantic prompt filled with detail. This came from thinking about Python functions, and how you can feed multiple functions into one another to create your desired output. I figured that this kind of mindset when creating prompts might be worth a shot, where each "function" was actually an individual Markdown file I created with its own specific purpose, that I could then feed into others to produce a "main" output, which in turn would be the prompt I could use to create that specific Polaroid.

<br>

I separated the "constant" elements of what we think of as a Polaroid photo (the aesthetic of the card, the format/dimensions, time period, etc) from the "variable" elements (the specific scene I chose, the location of the scene, the caption text, important details the Polaroid NEEDS for that scene, etc). I then made a reusable Markdown template file with this in mind (`Polaroid Creation Template Prompt.md`), whereby the "constant" elements of the prompt were included in detail, however the "variable" elements had `[REPLACE TEXT HERE]` as placeholder text.

<br>

To work in conjunction with this, I made another Markdown file (`15 Images and Potential Alternate Choices.md`) which contained a table of information for each of the 15 Polaroids which were going to be created, including details such as the polaroid number, the location, the caption text, and information about that actual moment I wanted to be captured. I then gave ChatGPT the 2 Markdown files, and gave it this single prompt:

<br>

```
Using both these files, please take my `Polaroid Creation Template Prompt.md` template file, and turn it into a copy/paste code block for `Number X` in the `## Main Polaroid Image Sequence` table within the `15 Images and Potential Alternate Choices.md` file. Anywhere you see `[REPLACE TEXT HERE]` within the `Polaroid Creation Template Prompt.md` template file I've provided, you replace that text with text relating to the scene highlighted in `Number X` in the `## Main Polaroid Image Sequence` table.
```

<br>

For each individual Polaroid, I created a new chat, pasted the above prompt in, and replaced the `X` in `Number X` with the number of the actual Polaroid I wanted ChatGPT to create the individual prompt template for. This meant that I was able to produce 15 individual prompts, ready to generate the Polaroid for me simply by copy/pasting, while ensuring that output remained consistent between each of the 15 images generated.

<br>

The prompt below is representative of the final template in use. It's the complete prompt, created by the method above, which was used to generate Polaroid #8: *(8) Mosquito Hunting by Candlelight and Slipper*.

<br>

```
Create a polaroid for my 19th-century Polaroid project. You are generating this Polaroid image from text, NOT editing an attached image.

Be sure to create the Polaroid frame, border, caption area, handwriting, paper texture, and surrounding background.

Initial instructions:
Please refer to the file `MSGen13-2.txt` if you need further understanding of the actual scene itself, as this `MSGen13-2.txt` file is the entire journal MacGeorge wrote, which was digitised by my university's archivists.

Scene:
Inside the Hotel Royal in Nice in the 1860s, late at night, a small Victorian hotel bedroom is lit mainly by candlelight. A tired traveller is comically hunting mosquitoes with a slipper, standing near the bed with the slipper raised toward a pale wall or ceiling. The scene should feel funny, domestic, human, and slightly exasperated rather than dramatic: rumpled bedding, travel trunks, folded clothes, a candleholder, shadows on the wall, and a few tiny mosquitoes visible near the candlelight or resting on the wall. It should look like a private, accidental snapshot of hotel life, not a staged portrait.

Location:
Hotel Royal, Nice, France.

Historical period:
1860s

Visual style:
A candid travel photograph, but imagined as if a traveller from the 1860s had access to a primitive instant camera. The image should look historically plausible, informal, observational, and not studio-posed.

Caption text:
(8) Mosquito Hunting by Candlelight and Slipper

Colour profile:
Muted antique colour, lightly faded, slightly desaturated, with natural colour still visible. More colour than sepia. Avoid strong orange/brown sepia!

Composition:
Natural documentary-style framing. Keep enough space around the traveller, bed, candle, wall, and surrounding hotel-room details so nothing important is cropped. The action should be readable at a glance: candlelight, slipper, mosquitoes, and comic late-night frustration.

The polaroid should have the following details:
- aged antique Polaroid paper
- lightly tattered and creased border
- cursive handwritten caption style (in ink)
- muted antique colour profile
- soft, slightly faded, slightly desaturated look
- "more colour than sepia" treatment

Polaroid shape requirements:
- use a classic portrait Polaroid print shape, not a square card
- show the entire Polaroid paper object, with no cropped or clipped edges
- make the outer frame taller than it is wide, with a noticeably thicker bottom caption border
- keep a rectangular inner photograph area above the caption space
- leave visible background/padding around all sides of the Polaroid

Must include:
- a candlelit 1860s hotel bedroom or private hotel room
- a Victorian traveller hunting mosquitoes with a slipper
- a few small visible mosquitoes or mosquito-like specks near the wall or candlelight
- rumpled bedding, travel luggage, and ordinary hotel-room clutter
- a humorous, tired, human mood rather than horror or slapstick

Avoid:
- modern objects
- modern clothing
- polished cinematic lighting
- overly dramatic fantasy style
- strong sepia
- text or writing inside the image

IMPORTANT:
This should look like a candid late-1860s travel snapshot, not a posed studio photograph.
```

<br>

### 3. ChatGPT Image Generation
I chose ChatGPT for image generation in this project as it produces strong results when given detail and stylistically specific prompts. I gave Google Gemini's image generation model a try too, however this didn't produce as photo-realistic results as I noticed with ChatGPT, altogether they looked slightly more "cartoonish"/CG than the results with ChatGPT. Another factor in this decision was that seeing as ChatGPT was aware of my project through the prior instructions I gave it in regards to the template creation, I felt more confident in it creating more accurate output, as opposed to just giving the individual Polaroid prompt (like the above prompt for Polaroid #8) to something like Gemini which would effectively be creating the image with no prior knowledge at all.

<br>

### 4. Caption Collaboration
I wanted to pair each Polaroid with a caption in the style of something you'd see at a museum on a placard alongside a photograph or painting. These captions were written collaboratively with ChatGPT in a human-led iterative process, whereby I requested the caption text to be taken from the *exact* journal text describing the scene or moment the Polaroid was created with, which was then reviewed, then edited however many times until the final draft. As such, the captions are a product of the collaborative drafting process between human x AI, rather than being purely human or purely AI.

## Methodology Approach Choices
AI makes it possible to generate images of moments that could once upon a time only be imagined. While the human role as a curator in selecting specific moments, ensuring prompts are grounded in legitimacy (and not AI hallucinations), and analytically evaluating every stage ensures that the outputs are not just a bunch of random AI generated images, but a carefully crafted collection of moments in time which were once experienced by the humans detailed within MacGeorge's journal.

# **The Polaroid Collection**
The fifteen Polaroids below present everday moments that we ourselves might take our phones out to snap pictures of now, but during the time of MacGeorge's journey would have required sufficient time to set up a camera to capture, thus making scenes like these much less common to see in old photography from that period. The captions below were produced through a collaborative drafting process between myself and ChatGPT. The conceptual framing and final editorial judgement on all caption text is my own.

Each Polaroid is paired with a caption regarding the scene itself, and are arranged in the order MacGeorge travelled (Glasgow to Rome).

| <img src="images/polaroid-1.png" width="1000"> | **(1) The Packing Chaos Before Departure** <br> *Glasgow — October 1867*  <br> <br> For weeks before departure, the MacGeorge household was consumed by the rituals of Victorian long-distance travel: boxes filled and emptied, cloaks bundled, umbrellas counted, and — somehow — a deal box containing six bottles of port loaded onto the coach. MacGeorge recorded the chaos in a quick pen-and-ink sketch, but a sketch can only freeze the finished pile; it cannot capture the deciding and re-deciding, the taking of far more than was necessary. In 1867, the camera was a formal instrument requiring advance preparation, a cooperative subject, and daylight — none of which the weeks before departure reliably provided. The anxiety of setting off belongs to every traveller, in every century. |
| <img src="images/polaroid-2.png" width="1000"> | **(2) Seasick Passengers on the Channel Crossing** <br> *English Channel — 30 October 1867*  <br> <br> The steamer *Albert Edwards* crossed to Boulogne carrying upwards of 200 passengers and, by MacGeorge's estimate, a corresponding number of basins. Seated by the funnel, he watched as "a great paleness settled into a thorough greenness of complexion" spread through the deck — a nurse in a fashionable bonnet, a dazed little boy in her charge, a newlywed bride going green while her husband regarded her with a "rueful expression." MacGeorge sketched some of these figures on the spot, but the sketch captures types, not the particular misery of the moment unfolding in real time. Victorian photography required a stationary subject, a clear sky, and several seconds of stillness — conditions the Channel crossing conspicuously failed to offer. |
| <img src="images/polaroid-3.png" width="1000"> | **(3) Coming Ashore with the Box of Port** <br> *Boulogne — 30 October 1867*  <br> <br> MacGeorge arrived in France encumbered: carpet bag in one hand, cloaks over one arm, and a deal box of port wine somewhere in the middle of it all, while Mrs MacGeorge and Margaret waited in an omnibus nearby. At the customs shed, the Douanier — apparently unimpressed by the imperial dignity of his role — waved him through with a gesture that seemed to suggest the Emperor could hardly be expected to trouble himself with duty on so minor an importation. It was a small comedy of bureaucratic indifference, lasting perhaps a minute, and entirely unrecordable by any camera then in existence. MacGeorge sketched the scene of coming ashore; what the sketch cannot render is the exchange. |
| <img src="images/polaroid-4.png" width="1000"> | **(4) The Moustached Traveller** <br> *Paris, Gare de Lyon — 4 November 1867*  <br> <br> On the departure platform in Paris, MacGeorge noticed a fellow passenger whose moustache was, he noted carefully, "not in the slightest degree exaggerated" — a sandy grey arrangement of extraordinary proportions, worn by a man accompanied by a party of ladies and apparently oblivious to the attention he attracted. MacGeorge drew him: a pen-and-ink profile that has survived in the journal for over 150 years. But drawing requires eye contact, time, and some cooperation from the subject; photography in 1867 required all of this and more. The stranger on the platform was never formally photographed and never knew he was observed. He exists only because someone was watching, and had a pencil. |
| <img src="images/polaroid-5.png" width="1000"> | **(5) The Locked Salle d’attente** <br> *Lyon station — 6 November 1867*  <br> <br> French railway practice baffled MacGeorge: passengers were locked into the waiting room until the train was ready, then released all at once onto the platform "like a set of maniacs, every one tumbling over the other in the race to get seats — families and friends separated." He contrasted this furiously with the orderly arrangements at Carlisle, where passengers boarded at leisure. A rough pencil sketch in the journal captures the surging crowd, but it is necessarily static — the whole point of the scene was its chaotic movement, the pushing and shouting and separating of parties. No camera of 1867 could have caught it either. Some moments resist capture by design. |
| <img src="images/polaroid-6.png" width="1000"> | **(6) Train-Carriage Lunch after Toulon** <br> *Railway between Toulon and Cannes — 7 November 1867*  <br> <br> Somewhere on the line between Toulon and Cannes, with the Mediterranean beginning to appear alongside the track, MacGeorge unwrapped a portion of roast fowl, helped himself to grapes and roasted chestnuts, and poured a glass of port. He recorded it in a single sentence. He did not sketch it. No photograph was taken. The scene left no visual trace of any kind — and yet this small, companionable meal, eaten in a rattling third-class carriage with the south of France sliding past the window, is one of the moments this project sets out to recover. It is ordinary in exactly the way that history tends to discard. |
| <img src="images/polaroid-7.png" width="1000"> | **(7) Moonlit Arrival and Hotel Trouble** <br> *Cannes — 7–8 November 1867*  <br> <br> The party arrived at Cannes in the evening, the moon approaching full, the town quiet and lit silver. After trying several hotels they settled on the Grand Hôtel de Nice, which MacGeorge declared would have been a perfectly good home "were it not for the insurmountable drawback of the evil smells which prevail in the corridors and which we found absolutely pestilential." The hotel, to its credit, acknowledged the problem and admitted an inability to resolve it. Victorian photography excelled at moonlit scenes and hotel façades; it had no mechanism for recording the smell of the corridors, nor the weary negotiation of arriving after dark and making do. MacGeorge's journal preserves both. |
| <img src="images/polaroid-8.png" width="1000"> | **(8) Mosquito Hunting by Candlelight and Slipper** <br> *Nice, Hotel Royal — November–December 1867*  <br> <br> Each night at the Hotel Royal followed the same ritual: candle in one hand, slipper in the other, a systematic inspection of walls and curtains. MacGeorge became an expert. Mosquitoes on flat surfaces were dispatched with the slipper — leaving, he noted drily, "a not very pleasant mark." Those hiding in curtains required a different technique: crushed in a towel between both hands. He drew himself mid-hunt, and the sketch survives. But the sketch is a pose; the real thing was a nightly domestic performance, carried out alone, by candlelight, in a foreign hotel room. Victorian portrait photography required subjects to sit still and look directly at the camera. Candlelight mosquito hunting was not that. |
| <img src="images/polaroid-9.png" width="1000"> | **(9) Band Music and Flower Girls** <br> *Nice, Jardin Public — 28 November 1867*  <br> <br> Three times a week — Tuesdays, Thursdays, Sundays — a band played in the Jardin Public, and girls moved through the crowd selling small bouquets of violets and roses for two sous. MacGeorge found it "novel to us to see ladies dressed in the fashions sitting in chairs in the open air under palm trees listening to music on the 28 of November." It was the ordinariness of it that struck him: the south of France simply continued as if late November were nothing unusual. The crowd, the flowers, the music, the Mediterranean light — all of it was technically within reach of a skilled Victorian photographer. What was not within reach was the informality of it, the happening of it, the sense that this was Tuesday and the band simply played. |
| <img src="images/polaroid-10.png" width="1000"> | **(10) La dame qui ne marche pas** <br> *Villefranche-sur-Mer — 19 November 1867*  <br> <br> When the party took a boat round the peninsula at Villefranche, Mrs MacGeorge stayed on board while MacGeorge and Mrs Ewing walked through the olive woods across the headland. One of the boatmen, accounting for her absence, described her as *"la dame qui ne marche pas"* — the lady who does not walk. MacGeorge records that the name "has since adhered to her," suggesting it became a private joke carried for the rest of the journey. Victorian photography could have produced a portrait of Mrs MacGeorge; it could not have caught the moment a stranger invented a nickname that stuck. That requires someone present, paying attention, and writing it down. |
| <img src="images/polaroid-11.png" width="1000"> | **(11) The Boy Picking Twelve Small Oranges** <br> *Villefranche-sur-Mer — 19 November 1867*  <br> <br> In a cottage garden near the shore, MacGeorge stopped and paid three sous for a dozen small oranges, watching a boy climb the tree to pick the ripest ones. The transaction complete, he tasted one: not very good — not yet enough sun. He noted that the oranges of this region were really best left on the tree until August, but that sellers could not afford to wait. It is perhaps the quietest moment in the entire journal: a small purchase, a brief exchange, an honest verdict. Nothing about it warranted a camera, and in 1867 no camera could have followed easily into a cottage garden for three sous' worth of fruit. It is preserved only because MacGeorge happened to write it down. |
| <img src="images/polaroid-12.png" width="1000"> | **(12) Roadside Lunch Beneath Èze** <br> *Èze — 22 November 1867*  <br> <br> The party stopped at Èze — a village on a rocky promontory, the Mediterranean a deep blue behind it — and ate lunch at the side of the road: cold fowl, country wine, a small bottle of port, tongue, and biscuits. MacGeorge made a sketch of the view. A photograph of Èze also exists, pasted into the journal — taken from a distance, showing the town against the cliff. The photograph shows the rock; the journal entry shows the lunch. Both were made on the same day, at the same location, by the same man. The gap between them is the gap this project explores: the photograph records what was worth formally documenting; the journal records what it was actually like to be there. |
| <img src="images/polaroid-13.png" width="1000"> | **(13) Snow in Nice with Roses** <br> *Nice — 3 December 1867*  <br> <br> On the morning of 3 December 1867, Nice woke to something its residents insisted was essentially impossible: snow that refused to melt. MacGeorge had been told repeatedly that if flakes ever fell in Nice, they dissolved on contact with the ground. They did not. The streets turned white, hailstorms followed, and local boys — surrounded by the rosebushes that still bloomed in December — did what children do everywhere with unexpected snow. A fellow guest at the Hotel Royal reported seeing two snow heaps in a nearby yard, each crowned with bunches of roses. MacGeorge noted it in his journal as a detail too strange and too local to leave out. |
| <img src="images/polaroid-14.png" width="1000"> | **(14) Blanchisseuses at the Paillon** <br> *Nice, mouth of the Paillon — 31 December 1867*  <br> <br> On the last day of 1867, MacGeorge walked to the mouth of the Paillon and found the smell "intolerable." The river served as the drainage channel for Nice's common sewers — and yet in this "foetid sink of pollution and filth," hundreds of washerwomen were at work, following their trade as they did every day. MacGeorge noted, with some relief, that the Hotel Royal had its laundry done elsewhere with clean water. The scene is saturated with the social history of the period: working women performing invisible labour in conditions that would have appalled the wealthy health tourists resting a short distance away. Formal Victorian photography documented the picturesque and the respectable. Neither word applied here. |
| <img src="images/polaroid-15.png" width="1000"> | **(15) In Rome at Last** <br> *Rome — 30 January 1868*  <br> <br> After three months on the road — Glasgow, London, the Channel, Boulogne, Paris, Lyon, Marseille, Cannes, Nice, Genoa, Pisa — MacGeorge arrived in Rome at last. The first impression was not what the Grand Tour demanded: dirty, narrow streets, an atmosphere "not calculated to excite enthusiasm," and, inevitably, the smell. "One must suspend judgment till daylight," he wrote, and went to bed. Victorian photography had produced thousands of images of Rome: the Forum, the Colosseum, the Palatine Hill in clear afternoon light. None of them recorded the experience of arriving exhausted after three months of travel, in the dark, through streets that smelled of drains. That — the real arrival — is what the journal alone preserves. |
